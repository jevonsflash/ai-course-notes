# Pandas Series对象介绍

Series 是 pandas *一维带标签数组，由一组值 + 对应的索引（index）组成。
底层存储数值的部分，就是 numpy 的 ndarray


```python

# 从列表创建Series
s1 = pd.Series([1, 3, 5, np.nan, 6, 8])
print("从列表创建的Series:")
print(s1)

# 从字典创建Series
s2 = pd.Series({'a': 1, 'b': 2, 'c': 3})
print("\n从字典创建的Series:")
print(s2)

# 指定索引
s3 = pd.Series([10, 20, 30], index=['x', 'y', 'z'])
print("\n指定索引的Series:")
print(s3)

```

```python

# Series操作
print("\nSeries操作:")
print("值:", s3.values)
print("索引:", s3.index)
```

## Pandas DataFrame对象介绍

`data_list`：原始数据，嵌套列表，每一个子列表代表**一行**

`columns`：列标签（表头），顺序和数据一一对应：姓名、年龄、城市、工资

`index`：行索引（行标签），这里自定义行标记：s1、s2、s3、s4，不再是默认 0,1,2,3

`describe()`：**数值型列的统计汇总**（只自动识别数字列：年龄、工资；文本列【姓名、城市】会自动忽略）

```python
# 创建示例DataFrame
data_list = [
    ['张三', 25, '北京', 5000],
    ['李四', 30, '上海', 7000],
    ['王五', 35, '广州', 6000],
    ['赵六', 28, '深圳', 8000]
]
df = pd.DataFrame(data_list, columns=['姓名', '年龄', '城市', '工资'], index=["s1", "s2", "s3", "s4"])
print(df)
# 查看数据
print(df.head(3))
print(df.tail(2))
print(df.describe())
```

## Pandas DataFrame查看数据

```python
# 选择数据
print(df['姓名'])
print('\n')
print(df[['姓名', '年龄']])
```

```python

print("\n使用loc选择（标签索引）:")
print(df.loc['s1'])  # 选择一行
print(df.loc[['s1', 's3']])  # 选择多行
print(df.loc['s1':'s3'])  # 切片选择


# s1行，姓名列
print(df.loc['s1', '姓名'])
# s1~s3行，姓名、工资两列
print(df.loc['s1':'s3', ['姓名','工资']])
```

```python
print("\n使用iloc选择（位置索引）:")
print(df.iloc[0])  # 第一行
print(df.iloc[[0, 2]])  # 选择多行
print(df.iloc[0:3])  # 切片选择
print(df.iloc[0:3,:-1])  # 切片选择
```

## Pandas DataFrame 筛选和过滤数据

布尔索引筛选：`df['年龄']>30` 以及`技术部的员工`，生成布尔 Series，保留 True 对应的行

年龄大于30的员工:
0    False
1    False
2     True
3    False
4     True


技术部的员工:
0     True
1    False
2     True
3    False
4    False

然后布尔 Series 和 DataFrame **行索引对齐**，只保留布尔值为 True 对应的行，False 的行直接丢掉。

`df.query()`：只是语法糖，内部最终也会生成布尔 Series 去筛选。

基础语法：
1. 表达式写在**单引号 / 双引号**里面
2. **列名直接写**，不用 `df['列名']`
3. 比较运算符：`>  <  >=  <=  ==  !=`
4. 逻辑连接：
   - `and` 并且
   - `or` 或者
   - `not` 取反

```python

# 创建示例数据
df = pd.DataFrame({
    '姓名': ['张三', '李四', '王五', '赵六', '钱七'],
    '年龄': [25, 30, 35, 28, 32],
    '部门': ['技术部', '销售部', '技术部', '人事部', '销售部'],
    '工资': [5000, 7000, 6000, 5500, 7500]
})

print("原始数据:")
print(df)
```

```python
# 条件筛选
print("\n年龄大于30的员工:")
print(df[df['年龄'] > 30])
print("\n技术部的员工:")
print(df[df['部门'] == '技术部'])
print("\n使用query方法筛选:")
print(df.query('年龄 > 28 and 工资 < 7000'))
```

## Pandas DataFrame对象 处理缺失值


```python
df=pd.DataFrame([
[1.0,5.0,9],
[2.0,np.nan,10],
[np.nan,np.nan,11],
[4.0,8.0,12],
[np.nan,np.nan,np.nan]],columns=['A', 'B', 'C'])

print("原始数据:")
print(df)
```

```python
# 检查缺失值
print("\n缺失矩阵:")
print(df.isnull())

print("\n缺失值统计:")
print(df.isnull().sum())

print("\n缺失矩阵:")
print(df.notnull())

print("\n非缺失值统计:")
print(df.notnull().sum())
```

```python


# 删除缺失值
print("\n删除包含缺失值的行:")
print(df.dropna())

print("\n删除全部为缺失值的列:")
print(df.dropna(how='all'))


```

```python
# 填充缺失值
print("\n用指定值填充缺失值:")
print(df.fillna(0))

print("\n用均值填充:")
print(df.fillna(df.mean()))
```
